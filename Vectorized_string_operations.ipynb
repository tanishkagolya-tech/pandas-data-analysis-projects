Vectorized string operations are usually applied on a Series (column) using .str.

```python
df['Name'].str.lower()
```

 Vectorized string operations are primarily applied on Pandas Series using .str. For DataFrames, you first select one or more string columns and then apply the string operation column-wise.

```python
df['Name'].str.upper()
```

```python
df['Name'].str.capitalize() #only starting word of line start with capital letter
```

```python
df['Name'].str.title() # each word start with capital letter
```

```python
#len
df['Name'][df['Name'].str.len()==82].values[0]
```

```python
df['Name'][df['Name'].str.len()]
```

```python
# strip
"                   tanishka                             ".strip()
```

```python
df['Name'].str.strip()
```

```python
# split -> get
df['lastname'] = df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df[['title','firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
df.head()

df['title'].value_counts()
```

```python
# replace
df['title'] = df['title'].str.replace('Ms.','Miss.')
df['title'] = df['title'].str.replace('Mlle.','Miss.')
```

```python
df['title'].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['firstname'].str.endswith('A')]
```

```python
# isdigit/isalpha.
df[df['firstname'].str.isdigit()]
```

```python
# applying regex
# contains
# search john - both case
df[df['firstname'].str.contains('john',case=False)]
```

```python
# find lastnames with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].*[aeiouAEIOU]$', regex=True)]
```

```python
# find lastnames with start and end char consonant
df[df['lastname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```

^[aeiouAEIOU] → starts with vowel

.* → anything in between

[aeiouAEIOU]$ → ends with vowel
